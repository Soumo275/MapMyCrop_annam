## 1. Background

**Q:** What is the background of the problem?

**A:** Accurate and timely information on **crop sowing, harvesting, acreage, and yield** remains a major challenge for central and state agriculture ministries. Current monitoring methods rely heavily on manual field visits or outdated models, making them time-consuming and error-prone. National-level models struggle with India's crop diversity, spectral overlaps, and regional heterogeneity, and often lack sowing/harvesting tracking capabilities. This highlights the need for a robust, AI-driven solution that leverages satellite remote sensing for near real-time agricultural monitoring.


## 2. Problem

**Q:** What is the problem being solved?

**A:** The problem being addressed is the lack of a reliable system for real-time monitoring of crop sowing, harvesting, acreage, and yield. Existing approaches are either manual or outdated, resulting in delays and inaccuracies in agricultural decision-making. Current national models fail to capture crop-specific variations across regions and do not provide sowing/harvesting timelines for the current year. The solution aims to bridge this gap using AI and satellite data to deliver accurate, timely, and crop-specific agricultural insights. This model acreage output will further help for immediate settlement in crop insurance cases.  

## 3. Customer

**Q:** Who is the solution for? Is it a user or a beneficiary?

**A:** The solution is intended for agents of both **government and private sectors** involved in **agricultural monitoring and planning**. It can be used as input in **Yield Estimation System using Technology**, a system used in India's Pradhan Mantri Fasal Bima Yojana **(PMFBY)** by national and state level agri-tech service providers, agricultural officers, and crop insurance agents. These users will interact directly with the system to access timely and accurate crop data. While they are the primary users, the ultimate beneficiaries include **farmers, policymakers, and food supply chains** that rely on these insights for timely action and support.

## 4. Value Proposition

**Q:** Why does this problem need to be solved?

**A:** This problem needs to be solved because current methods for monitoring crop sowing, harvesting, acreage, and yield are **manual, outdated, and often inaccurate**. They fail to capture real-time changes and crop-specific variations, which hampers agricultural planning, policy-making, and food security. An AI-based system using satellite data can provide timely, accurate, and scalable insights for better decision-making about varieties of crop of current year.

## 5. Product

**Q:** What does the solution look like (from a user experience perspective)?

**A:** From a user experience perspective, the solution is a user-friendly **web application and a JAVA-based desktop tool**. Users can upload satellite remote sensing images (GeoTIFFs/Generic Binary), view crop maps, monitor sowing and harvesting events, and estimate crop acreage and yield. The interface displays mapped areas, timelines, and area statistics interactively, making complex AI-driven analysis accessible to non-technical users like farmers, planners, and government officials. This works with very small number of ground geo-tagged points and can provide specific crop acreage and  yield estimation in  a short time.

## 7. Risks and Challenges

**Q:** What challenges may be encountered and how can they be mitigated?

**A:** Major challenges include **spectral overlap between crops**, **temporal variability**, and **limited ground-truth data**. These are mitigated through advanced hybrid deep learning 1D-CNN model, which capture both spatial and temporal dynamics. The problem of heterogeneity within a class is addressed using the innovative **"individual sample as mean" (ISM)** training parameter approach, where each sample is fed into the model individually, rather than taking the mean of the samples.  A powerful web application streamlines data upload and visualisation, while a Java-based yield estimator empowers users with rapid, reliable insights, minimising dependence on manual methods. Model is scalable across large number of various crops as well as covering district level area size.

## 11. Data

**Q:** What kind of data is needed, and how will it be collected (both training and ongoing)?

**A:** **Temporal Satellite Data**: Multi-date satellite imagery (e.g., PlanetScope/Sentinel 2/ etc.) is processed into temporal vegetation indices such as NDVI and MSAVI2, capturing crop growth patterns over time. If optical remote sensing data is partially available or unavailable then SAR temporal images can be used.

**Geo-tagged Training and Testing Data**: Collected through field visits with labelled samples for both target and non-target classes, stored in generic binary format for integration into ML/DL pipelines with the help of mobile applications.

**Yield Estimation Data**: Includes sample location-specific yield values from field observations, also stored in generic binary format, to support future regression or yield estimation modules of complete study area.

## 13. Acievements

**Q:** What were the key achievevemts of the project?

**A:** Three test sites were considered for testing the model to incorparte the diffenret locations and types of crops. The three sites with the key findings are as follows:

1.   Patiala site for mapping paddy crop **transplantation**, **harvesting** and paddy **stubble burnt **fields.  DL model output with high accuracy have been achieved for all three cases.

2.   Jhansi site for **Pea** and **Wheat** crop sowing monitoring and total acreage estimation. DL model has shown high accuracy results for green pea crop as well.

3.   Narmadapuram site for **Moong**/ Green gram Pulse crop sowing monitoring, total crop acreage and yield estimation. This site has also been scaled up to achieve **district** level **acreage** mapping as well as **yield** estimation.
In Narmadapuram effect of change in government policy about providing canal water during summer crop has also been shown. It can be seen by observing the increase in the summer crop area from 2004 to 2025 where the mapped summer crop area in 2004 was 12,686 Ha, 53,907 Ha in 2017 and finally in 2025 the area increased to 96,176 Ha for only Moong Pulse crop.In 2025 only the moong pulse crop acreage have been mapped.

We have also built a **web application** where we can integrate the DL model along with the JAVA based software for visualising the area of interest, mapping the crops of interest and estimating the acreage and yield in future.
